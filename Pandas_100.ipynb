### Q4. Perform arithmetic operations on a Series.
> 💡 **Hint:** Use `+`, `-`, `*`, `/`, `//`, `%`, `**` directly on the Series. Operations are **element-wise** and **index-aligned**.

```python
# it creates new series original unchanged 
s = pd.Series([10,20,30,40])

print("original series:")
print(s)

print("Addition:")
print(s+5)

print("subtraction:")
print(s-5)

print("multiplication:")
print(s*2)


print("division:")
print(s/2)

print("floor division:")
print(s // 3)

print("modulus:")
print(s % 3)

print("power:")
print(s ** 2)
```

### Q5. Find mean, min, max of a Series.
> 💡 **Hint:** Use `.mean()`, `.min()`, `.max()`, `.sum()`, `.median()`, `.std()` directly on the Series.

```python
s = pd.Series([10,20,30,40])
print("mean:",s.mean())
print("min:",s.min())
print("max:",s.max())
print("sum:",s.sum())
print("median:",s.median())
print("std:",s.std())
```

### Q6. Apply conditional filtering on a Series.
> 💡 **Hint:** Use a boolean mask: `s[s > 50]`. You can also chain conditions with `&` and `|

```python
s = pd.Series([15,35,65,20,40])

print(s[s>20])
print(s[(s > 20) & (s < 60)])
print(s[(s < 20) | (s >50)])
```


### Q56. Detect outliers in a column.
> 💡 **Hint:** Use the **IQR method**: compute Q1, Q3, IQR = Q3–Q1. Outliers are values below `Q1 - 1.5*IQR` or above `Q3 + 1.5*IQR`. Alternatively use Z-score: `|z| > 3`.

```python
df = pd.DataFrame({
    'salary':[30,35,40,45,50,55,200]
})

q1 = df['salary'].quantile(0.25)
q3 = df['salary'].quantile(0.75)

iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

print(lower,upper)

ouliers = df[(df['salary'] < lower) | (df['salary'] > upper)]

print(ouliers)
```

### Q72. Chain multiple operations.
> 💡 **Hint:** Use method chaining with `(df.copy().loc[...].sort_values(...)[cols].assign(...).reset_index(drop=True))` — wrap in parentheses for multi-line readability.

```python

df = pd.DataFrame({
    "Name": ["John", "Alice", "Mary", "Bob"],
    "Age": [24, 32, 25, 40],
    "Salary": [50000, 70000, 60000, 80000]
})

result = (
    df.copy().loc[lambda x:x['Age'] >25]
    .sort_values('Salary',ascending=False)
    [['Name','Age','Salary']].assign(bonus = lambda x:x['Salary'] * 0.10)
    .reset_index(drop = True)
)
print(result)
```

### Q73. Normalize numeric columns (Min-Max).
> 💡 **Hint:** `(df["col"] - df["col"].min()) / (df["col"].max() - df["col"].min())` — result is in range [0, 1].

```python

df = pd.DataFrame({
    "Marks": [40, 60, 80, 100]
})
df['normalized'] = ((df['Marks'] - df['Marks'].min()) /
                    (df['Marks'].max() - df['Marks'].min()))
print(df)
```

### Q74. Scale numerical data (Z-score standardization).
> 💡 **Hint:** `(df["col"] - df["col"].mean()) / df["col"].std()` — result has mean=0 and std=1.

```python
df = pd.DataFrame({
    "Marks": [40, 60, 80, 100]
})

df['zscore'] = (df['Marks'] - df['Marks'].mean()) / df['Marks'].std()
print(df)
print(df['zscore'].mean())
print(df['zscore'].std())
```
