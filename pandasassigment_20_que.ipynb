```python
# 7. Use loc to select:
#    - A single row
#    - Multiple rows
#    - Conditional rows
```

```python
student = pd.DataFrame({
    'Name':['John','Mary','Alice','Bob'],
    'subject':["english","maths","science","hindi"],
    'fees':[50000,60000,55000,70000]
}, index=['E1','E2','E3','E4'])

print(student)
```

```python
student.loc['E1']
student.loc[['E1','E3'], ['Name','fees']]
student.loc[student['fees'] < 70000]
```
